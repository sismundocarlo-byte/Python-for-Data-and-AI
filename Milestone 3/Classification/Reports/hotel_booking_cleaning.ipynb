## 3. Remove Duplicates

## 4. Remove columns that can cause target leakage

```python
# Step 4: Remove columns that could cause target leakage

# These columns may contain information that would not
# be available when we first predict a booking cancellation.
leaky_columns = [
    "reservation_status",
    "reservation_status_date",
    "assigned_room_type"
]

# Keep only the columns that actually exist in our dataset.
columns_to_drop = []

for column in leaky_columns:
    if column in df.columns:
        columns_to_drop.append(column)

# Display the columns that will be removed.
print("Columns to remove:", columns_to_drop)

# Remove the selected columns from the dataset.
df = df.drop(columns=columns_to_drop)

# Check the remaining columns.
print("Remaining number of columns:", df.shape[1])
print("Remaining columns:")
print(df.columns.tolist())
```

## 5. Handle missing values

```python

# STEP 5: HANDLE MISSING VALUES

# 5.1. Handle missing values in the "children" column
if "children" in df.columns:
    # Replace missing children values with 0
    df["children"] = df["children"].fillna(0)


# 5.2. Handle missing values in the "country" column
if "country" in df.columns:
    # Replace missing countries with "Unknown"
    df["country"] = df["country"].fillna("Unknown")


# 5.3. Handle missing values in the "agent" column
if "agent" in df.columns:

    # Create a new column to identify bookings
    # that have an agent ID recorded.
    df["has_agent"] = df["agent"].notna().astype(int)

    # Replace missing agent IDs with 0.
    # Convert IDs to integers first, then strings,
    # so the column can be treated as categorical data.
    df["agent"] = df["agent"].fillna(0).astype(int).astype(str)


# 5.4. Handle missing values in the "company" column
if "company" in df.columns:

    # Create a new column to indicate whether
    # a company ID was recorded.
    df["has_company"] = df["company"].notna().astype(int)

    # Remove the original company ID column
    # because it contains too many missing values.
    df = df.drop(columns=["company"])


# 5.5. Check whether any missing values remain
print("Missing values in each column:")

missing_values = df.isna().sum()

# Display only columns that still have missing values
print(missing_values[missing_values > 0].sort_values(ascending=False))
```

## 6. Standardize the meal category

```python

# STEP 6: STANDARDIZE THE MEAL CATEGORY

# Check whether the "meal" column exists in the dataset
if "meal" in df.columns:

    # Display the meal categories before making changes
    print("Meal categories before cleaning:")
    print(sorted(df["meal"].dropna().unique()))

    # Replace "Undefined" with "SC"
    # Both labels represent no meal package in this dataset.
    df["meal"] = df["meal"].replace("Undefined", "SC")

    # Display the meal categories after making changes
    print("\nMeal categories after cleaning:")
    print(sorted(df["meal"].dropna().unique()))
```

## 7. Filter impossible or extreme records

```python

# STEP 7: FILTER IMPOSSIBLE OR EXTREME RECORDS

# Record how many rows we have before filtering
rows_before_filtering = len(df)

# 7.1. Calculate the total number of guests for each booking
total_guests = df["adults"] + df["children"] + df["babies"]

# 7.2. Calculate the total number of nights for each booking
total_nights = (
    df["stays_in_weekend_nights"]
    + df["stays_in_week_nights"]
)

# 7.3. Check whether each booking meets our cleaning rules

# A booking must have at least one guest
has_guests = total_guests > 0

# A booking must be for at least one night
has_nights = total_nights > 0

# The average daily rate (ADR) must not be negative
valid_adr_minimum = df["adr"] >= 0

# The ADR must not be greater than 1000
valid_adr_maximum = df["adr"] <= 1000

# Keep only rows that satisfy ALL four conditions
valid_rows = (
    has_guests
    & has_nights
    & valid_adr_minimum
    & valid_adr_maximum
)

# Count how many rows will be removed
rows_to_remove = (~valid_rows).sum()

print("Rows before filtering:", rows_before_filtering)
print("Rows excluded by cleaning rules:", int(rows_to_remove))

# Remove rows that do not meet the rules
df = df.loc[valid_rows].copy()

# Record the number of rows after filtering
rows_after_filtering = len(df)

print("Rows remaining after filtering:", rows_after_filtering)
print("Total rows removed:", rows_before_filtering - rows_after_filtering)
```

## 8. Engineer new features

Feature engineering creates useful fields from existing columns:

- `total_nights`: weekend nights + weekday nights.
- `total_guests`: adults + children + babies.
- `has_kids`: 1 if children or babies are present, otherwise 0.
- `arrival_month`: month name converted to a number from 1 to 12.
- `arrival_weekday`: day of week, where Monday = 0 and Sunday = 6.
- `adr_per_guest`: average daily rate divided by total guests.
- `has_prev_cancel`: 1 if the customer has any previous cancellations, otherwise 0.

Arrival year and day-of-month are dropped in this project: year may not generalize to future years, while day-of-month is omitted as a granular feature. The month number is retained. Feature usefulness should ultimately be checked with validation and domain knowledge.

```python

# STEP 8: FEATURE ENGINEERING

# 8.1. Calculate the total number of nights for each booking
df["total_nights"] = (
    df["stays_in_weekend_nights"]
    + df["stays_in_week_nights"]
)


# 8.2. Calculate the total number of guests
df["total_guests"] = (
    df["adults"]
    + df["children"]
    + df["babies"]
)


# 8.3. Check whether the booking includes children or babies
# If children or babies are present, the result is 1.
# Otherwise, the result is 0.
df["has_kids"] = (
    (df["children"] + df["babies"]) > 0
).astype(int)


# 8.4. Convert the arrival month name into a number
# January = 1, February = 2, ..., December = 12
month_numbers = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

df["arrival_month"] = df["arrival_date_month"].map(month_numbers)

# Check whether any month names were not recognized
if df["arrival_month"].isna().any():
    print("Warning: Some arrival months were not recognized.")
    print(df.loc[df["arrival_month"].isna(), "arrival_date_month"].unique())


# 8.5. Create a proper arrival date
# Combine the year, month number, and day into one date.
arrival_dates = pd.to_datetime({
    "year": df["arrival_date_year"],
    "month": df["arrival_month"],
    "day": df["arrival_date_day_of_month"]
}, errors="coerce")

# Check whether all arrival dates are valid
if arrival_dates.isna().any():
    raise ValueError("Some arrival dates are invalid. Please check the data.")

# Extract the day of the week
# Monday = 0, Tuesday = 1, ..., Sunday = 6
df["arrival_weekday"] = arrival_dates.dt.dayofweek


# 8.6. Calculate the average daily rate per guest
# Avoid dividing by zero if a booking has no guests.
df["adr_per_guest"] = (
    df["adr"] / df["total_guests"].replace(0, pd.NA)
)


# 8.7. Check whether the customer had previous cancellations
# 1 = at least one previous cancellation
# 0 = no previous cancellations
df["has_prev_cancel"] = (
    df["previous_cancellations"] > 0
).astype(int)


# 8.8. Remove the original arrival date columns
# because we created new date-related features.
df = df.drop(columns=[
    "arrival_date_year",
    "arrival_date_month",
    "arrival_date_day_of_month"
])


# 8.9. Display the new features to check the results
print("Feature engineering complete!")

new_features = [
    "total_nights",
    "total_guests",
    "has_kids",
    "arrival_month",
    "arrival_weekday",
    "adr_per_guest",
    "has_prev_cancel"
]

print("New features created:")
print(new_features)

display(df[new_features].head())
```
