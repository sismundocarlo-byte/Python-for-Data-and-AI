```python

from datetime import datetime, date, timedelta             # this line of code get the module for time and date in python.

today_date = date.today()                                  # store the current date in a variable called today_date.

project_title = input("Please enter the project title: ")  # user input  will be title of our project.


while True:                                                # loop until the user enters a valid date format
    try:
        project_deadline = datetime.strptime (             #datetime module convertion to python  
            input("Please enter the project deadline (YYYY-MM-DD): "), "%Y-%m-%d"
        ).date()
        break                                              # Exit the Loop if the input is valid
    except ValueError:
        print("Invalid date format")
        print("Please enter the date in the format YYYY-MM-DD")

days_remaining = (project_deadline - today_date).days        # Answer is converted to integers

print ("/n" + "=" * 50)
print ("         PROJECT DEADLINE CHECKER")
print ("/n" + "=" * 50)
print ("Project Title:", project_title)
print ("Today Date:", today_date)


if days_remaining < 0:
    print ("Your Project Status: Overdue ")
    print ("Project already passed for", days_remaining, " days")
elif days_remaining >= 30 :
    print ("Your project status : Plenty of time")
    print ("Your project still have", days_remaining, " days")
elif days_remaining == 3 :
    print ("Urgent Reminder Your project will be on deadline in 3 days")
elif days_remaining < 30 and days_remaining > 3 :
    print ("Warning your project is less than a month")   
    print ("You only have", days_remaining, "days in your project before dealline")
else :
    print ("Hey Lazy Kid, Today is Your Deadline")

print("/n" + "=" * 50)
    
    
   
```
